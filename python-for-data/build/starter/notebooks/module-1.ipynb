### Try it: the lesson example
Run this cell. It adds up the sales on 1 January with plain Python.

```python
import csv
total = 0
orders = 0
with open("data/cafe_sales_2025.csv", newline="") as file:
    for row in csv.DictReader(file):
        if row["date"] != "01/01/2025":
            continue
        if row["quantity"] == "":
            continue
        price = float(row["unit_price"].replace("$", ""))
        total += int(row["quantity"]) * price
        orders += 1
print(f"1 January: {orders} order lines, ${total:.2f}")
```
