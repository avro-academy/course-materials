### Exercise 2: explore
1. Run `df.info()`. Which column has missing values?
2. How many order lines did each staff member serve? (`value_counts`)
3. Run `df["item"].unique()`. How many ways is "Flat white" written?

```python
import pandas as pd
df = pd.read_csv("data/cafe_sales_2025.csv")
df.info()                               # quantity and payment have missing values
print(df["staff"].value_counts())
print(sorted(i for i in df["item"].unique() if "flat" in i.lower()))   # 4 ways: 'Flat White', 'Flat white', 'Flat white ', 'flat white'
```

### Exercise 3: filter
1. How many order lines were for a Long black with a quantity of 2 or more?
2. Show every order Priya served on 25/12/2025 (columns `time`, `item`, `quantity`).

```python
import pandas as pd
df = pd.read_csv("data/cafe_sales_2025.csv")
print(len(df[(df["item"] == "Long black") & (df["quantity"] >= 2)]))
print(df.loc[(df["staff"] == "Priya") & (df["date"] == "25/12/2025"), ["time", "item", "quantity"]])
```

### Exercise 4: sort and new columns
In `products`, add a column `margin_note` that is True when the cost price is under $1. Then sort the products from cheapest to most expensive to make.

```python
import pandas as pd
products = pd.read_csv("data/products.csv")
products["margin_note"] = products["cost_price"] < 1
print(products.sort_values("cost_price"))
```
